## 2. Phase 2 Verification Exit Code Breakdown

In SWE-bench, patches are evaluated by executing held-out verification tests. Rather than treating failures as a monolithic `Fail`, ATIF-Gemma classifies outcomes by exact Phase 2 pytest exit codes:
- **Exit Code 0**: Task resolved (unit tests pass).
- **Exit Code 1**: Unit test assertion failure (patch executes, but test fails).
- **Exit Code 2**: Pytest collection / import crash (syntax errors, missing packages, import shadowing).
- **Exit Code -1**: No patch submitted (budget or timeout exhaustion).


```python
ec_counts = df['test_exit_code'].value_counts().sort_index()
ec_df = pd.DataFrame({
    'Exit Code': ec_counts.index,
    'Meaning': [
        'Timeout / No Patch (-1)',
        'Resolved / Tests Pass (0)',
        'Assertion Failure (1)',
        'Collection / Import Error (2)'
    ],
    'Count': ec_counts.values,
    'Percentage (%)': (ec_counts.values / len(df) * 100).round(1)
})
display(ec_df)
print('Key Insight: 31.2% of agent runs failed due to collection/import crashes before test execution!')
```
